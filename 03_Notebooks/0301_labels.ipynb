```python
import pandas as pd
```

#### categorize nutrients

```python
nutrient_categories = pd.read_csv("../02_Data_formatted/subcategories.csv")

aminos = nutrient_categories.amino_acids.dropna().values

nv = pd.read_csv("../02_Data_formatted/nv_df.csv")

nutrient_categories = nutrient_categories.drop("fats", axis=1)
```

```python
label_conversion = {"amino_acids":"aa", "minerals":"m", 
                    "vitamins":"v"}
```

```python
label_dict = {}
for col in nutrient_categories.columns.values:
    for v in nutrient_categories.loc[:,col].dropna():
        label_dict[v] = label_conversion[col]
```

```python
categorized_cols = ["_".join([label_dict[v], v]) if v in label_dict.keys() else 
                    v for v in nv.columns.values]

nv_categorizednutrients = nv.copy(deep=True)

nv_categorizednutrients.columns = categorized_cols
nv_categorizednutrients.to_csv("../02_Data_formatted/0201_data/nv_categorized.csv")
```
